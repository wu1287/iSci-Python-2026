## Problem 2 — Simple calculator
Ask the user for two numbers and print:
- Sum, difference, product, quotient
- Floor division and remainder
- One power example (e.g., a**b)

```python
# TODO: Implement a small calculator using input(). Be sure to cast input to numbers (int or float).
# Hints:
# a = float(input("Enter a number: "))
# b = float(input("Enter another number: "))

#Prompts the user to enter a number
a = float(input("Enter a number"))

#Prompts the user to enter another number
b = float(input("Enter another number"))

#Printing the computed sum
print("Sum:", a, "+", b,"=", a+b)

#Printing the computed difference
print("Difference:", a, "-", b, "=", a-b)

#Printing the computed product
print("Product:", a, "*", b, "=", a*b)

#Printing the computed floor division and remainder
print("Floor division:", a, "÷", b, "= ", a//b, "Remainder:", a%b)

#Printing the exponent a^b
print("Exponent:", a, "^", b, "=", a**b)
```

## Problem 3 — Circle computations
Ask the user to input a radius `r`, then compute and output:
- Circumference $C = 2\pi r$
- Area $A = \pi r^2$

Use 3.14159 as an approximation for $\pi$.

```python
# TODO: Compute circle circumference and area.

#Prompts the user to enter a radius
r = float(input("Enter a radius:"))

#Prints the radius
print("Radius:", r)

#Printing the computed circumference
print("Circumference:", 2*3.14159*r)

#Printing the computed area
print("Area:", 3.14159*r**2)
```
