## Problem 1 — Sum of first N square integers
Ask the user for `N` and compute the sum $1^2 + 2^2 + ... + N^2$ using a loop.

```python
#Prompts the user to enter an integer
n = int(input("Enter a positive integer:"))

#Restricts the input to positive integers
if n < 0:
    print("Your integer must be positive.")

#Only run the rest of the code (for computating the sum) if the integer is not negative
else: 

    #Variable for our sum
    sum = 0

    #Runs the for loop until i equals n+1, as for loop stops at the previous number
    for i in range(n+1):
        #After every iteration of the loop, the new sum equals the old sum plus i squared
        sum += i**2

    #This series of if's and elif's allow the resulting printed number to be mathematically accurate
    if n == 0:
        #1^2, 2^2 and + ... is not necessary if n is 0
        print(f"0^2 equals {sum}.")
    elif n == 1:
        #2^2 is not neccessary if n is 1
        print(f"1^2 equals {sum}.")
    elif n == 2:
        #+ ... is not necessary if n is 2
        print(f"The sum of 1^2 + 2^2 is {sum}.")
    elif n == 3:
        #+ ... is not necessary if n is 3
        print(f"The sum of 1^2 + 2^2 + 3^2 is {sum}.")
    else:
        #+ ... is only required for integers 4 and above
        print(f"The sum of 1^2 + 2^2 + ... {n}^2 is {sum}.")

```

## Problem 2 — Multiplication table
Print a 10x10 multiplication table (rows 1..10, columns 1..10). Format it into aligned columns.

```python
# TODO: Implement a nested loop printing a multiplication table.
# Aesthetic Tip: use f-strings with field width (e.g., f"{value:4d}") for alignment (replace "value" with the actual integer you want to print).
# Concretely, you can use print(f"{value:d},end="") to keep printing on the same line (the 'end=""' prevents a new line from being started)
# To go to the next line, you can use print("")

#Defining the variables
row = 1
column = 1

#Title
print("        10x10 Multiplication Table:")


#Runs the loop until row 11 (instead of 10), as the for loop stops at the previous number
for row in range(1,11):

    #Runs the loop until column 11
    for column in range(1,11):

        #For every (row,column) positon, output their product. Use end"" to continue on the same line. Use :4d suffix for alignment.
        #This nested loop allows for every posititon to be included: (1,1) (1,2), ... (2,1) (2,2), ... (10,10)
        print(f"{row*column:4d}", end="")
        
        #Increase the column number for every interation of the loop
        column += 1

    #Starts a new line
    print("")
    #Increase the row number for every iteration of the loop
    row += 1
```

## Problem 3 — Guessing game (while loop)
Choose a secret number (hardcode or random). Repeatedly ask the user to guess until correct.
- If the guess is too low/high, print a hint.
- Count the number of guesses.

```python
# TODO: Implement a guessing game using a while loop.
# Optional: import random and use random.randint(1, 100)
import random

#Prompts the user to enter a number
n = int(input("Guess a number between 1 and 100:"))

#r is our secret number - a random number between 1 - 100
r = random.randint(1,100)

#i is our counter for the while loop
i = 1

#Runs the code below as long as the number guessed is not equal to the secret number
while n != r:
    #Use end="" to continue on the same line
    print("The number you guessed is not correct. ", end="")

    #Increase the counter of the while loop by one with each incorrect guess
    i += 1
    
    #Notifies user if the number is not in 1-100
    if n > 100 or n < 1:
        print(n, "is not between 1 and 100. Please guess a number between 1-100!")
        n = int(input("Guess another number between 1 and 100:"))

    #If number guessed is greater than r, prompts the user to guess another number.
    elif n > r:
        print(f"Your guess of {n} is too high.")
        n = int(input("Guess another number between 1 and 100:"))

    #If the number guessed is less than r (this is the only possible option since n is already not r), prompts the user to guess another number.   
    else:
        print(f"Your guess of {n} is too low.")
        n = int(input("Guess another number between 1 and 100:"))
    

#The while loop is exited when the correct number is guessed
print(f"You guessed the secret number! The number is {n}!")
#Outputs the counter i as the number of attempts
print(f"You took {i} tries to guess the secret number.")
```
