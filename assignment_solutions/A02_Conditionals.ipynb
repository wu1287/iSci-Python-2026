## Problem 1 — Sign & parity
Ask for an integer and print whether it is:
- Positive, Negative or Zero
- Even or Odd

```python
# TODO: Classify the integer by sign and parity.
# Tip: use % 2 to test even/odd.

#Prompts the user to enter a number
a = float(input("Enter a number:"))

#Outputs positive if the number is greater than 0.
if a > 0:
    print(a, "is positive.")

#Outputs 0 if the number is 0.   
if a == 0:
    print(a, "is zero.")

#Outputs negative if the number is less than 0.    
if a < 0:
    print(a, "is negative.")

#Outputs even if the number is divisible by 0.
if a % 2 == 0:
    print(a, "is even.")

#Otherwise output odd.    
else:
    print(a, "is odd.")
```

## Problem 2 
Given a float x, determine if it is in the interval $[0,1], (1,2], (2,3]$ or none of them.


```python
# TODO: Implement an if-elif-else 

#Prompts the user to enter a number
x = float(input("Enter a number:"))

#Outputs the interval [0,1] if the number is 0 ≤ x ≤ 1
if 0 <= x <= 1:
    print("Your number is in the interval [0,1].")
#Outputs the internal (1,2] if the number is 1 < x ≤ 2
elif 1 < x <= 2:
    print("Your number is in the interval (1,2].")

#Outputs the interval (2,3] if the number is 2 < x ≤ 3   
elif 2 < x <= 3:
    print("Your number is in the interval (2,3].")

#Otherwise (if all of the conditions above are not satisfied), the number is not in any of the intervals    
else:
    print("Your number is not in any of the listed intervals.")
```

## Problem 3 — Triangle Classifier
 Ask the user for three side lengths a, b, c.
 1. Check if they can form a valid triangle (all sides have positive length, and all triangle inequalities are satisfied).
 2. If valid, classify as:
    - Equilateral (all sides equal),
    - Isosceles (two sides equal),
    - Scalene (all sides different).

3. Otherwise, print that it's not a valid triangle.

Hint: you can place an `if`-block inside another `if`-block

```python
# TODO: Implement triangle classification.
# Example interaction:
# Enter side a: 3
# Enter side b: 3
# Enter side c: 3
# Output: "Equilateral triangle"
```

```python
#Prompts the user to enter 3 numbers
a = float(input("Enter a number (a triangle side length):"))
b = float(input("Enter another number:"))
c = float(input("Enter another number:"))

#Check if the side lengths make a valid triangle by checking if the sum of any 2 side lengths is larger than the remaining side
if a + b > c and b + c > a and a + c > b:
    #Output valid triangle 
    print(f"The side lengths {a}, {b}, {c} make a valid triangle.")

    #Check if all 3 side lengths are the same. If yes, output equilateral
    if a == b == c:
        print(f"The side lengths {a}, {b}, {c} make an equilateral triangle.")

    #Checking if any 2 of the side lengths are the same. If yes, output isosceles  
    elif a == b or b == c or a == c:
        print(f"The side lengths {a}, {b}, {c} make an isosceles triangle.")

    #Otherwise, all 3 side lengths must be different and output scalene.    
    else:
        print(f"The side lengths {a}, {b}, {c} make a scalene triangle.")

#Otherwise, the side lengths do not make a valid triangle.        
else:
    print("The side lengths you inputted does not make a valid triangle.")
```
