Write two functions:

c_to_f(celsius) converting °C to °F using the formula F = C * 9/5 + 32

f_to_c(fahrenheit) converting °F to °C using C = (F - 32) * 5/9

Test both on a few values.

```python
# TODO: Implement c_to_f and f_to_c

#print(c_to_f(0))   # -> 32
#print(f_to_c(212)) # -> 100


#Defining the c_to_f function with the input argument being celcius
def c_to_f(celsius):
    #Return an farenheit output value after calculation using formula
    return celsius * 9/5 + 32

#Defining the f_to_c function with the input arugment being farenheit
def f_to_c(farenheit):
    #Return an celcius value output after calculation using formula
    return (farenheit - 32) * 5/9

#Prompt user to input a number in C
a = float(input("Enter a temperature in Celcius:"))
#Print the value that is outputted via running the input through the c_to_f function
print(f"{a}˚C is {c_to_f(a)}˚F.")

#Prompt user to input a number in F
b = float(input("Enter a temperature in Farenheit:"))
#Print the value that is outputted via running the input through the f_to_c function
print(f"{b}˚F is {f_to_c(b)}˚C.")
```

## Problem 2 — BMI with keyword-only flag
Write `bmi(weight_kg, height_m, *, rounded=False)` and return BMI value, optionally rounded to 1 decimal place when `rounded=True`.
Test it with a couple examples, including weight_kg = 70, height_kg = 1.82 (once rounded and once without rounding).

Recall the formula for bmi: weight_kg/(height_m)^2.

```python
# TODO: Define bmi(...) and test with a couple examples, including weight_kg = 70, height_kg = 1.82. Do this once with rounding and once without rounding

#Defining the BMI function with the input arguments of weight and height. Rounded is optional as it is after *
def bmi(weight_kg, height_m, *, rounded = False):
    #Calculating BMI using the formula
    bmi_value = weight_kg/(height_m ** 2)
    
    #Check if the rounded argument is true, if yes, round the bmi_value to 1 decimal place
    if rounded == True:
        return round(bmi_value, 1)
    else: #otherwise return unrounded value
        return bmi_value

# Test with 70 and 1.82
weight_kg = 70
height_m = 1.82
print(f"Testing with example: The BMI at 70 kg and 1.82 m is {bmi(weight_kg, height_m)}. ", end="")
print(f"The BMI rounded to 1 decimal place is {bmi(weight_kg, height_m, rounded = True)}")

#Prompts user to input weight and height
kg = float(input("Enter a weight in kg:"))
m = float(input("Enter a height in m:"))

#Print the BMI via running the inputs through the BMI function
print(f"The BMI at {kg}kg, {m}m is {bmi(kg, m)}.")
#Print the rounded value by adding the argument rounded = True.
print(f"The BMI rounded to 1 decimal place is {(bmi(kg, m, rounded = True))}.")
```

## Problem 4 — Quadratic roots

```python
# TODO: Implement quad_roots(a, b, c) for real roots. Use math.sqrt for discriminant >= 0.
import math

#Defining the quad_roots functions with the inputs a,b,c
def quad_roots(a,b,c):
    #If discriminant is greater than 0, return both values obtained from the quadratic formula 
    if b**2 - 4*a*c > 0:
        return (-b + math.sqrt(b**2-4*a*c)) / (2*a), (-b - math.sqrt(b**2-4*a*c)) / (2*a)

    #If discriminant is 0, return the single solution
    elif b**2 - 4*a*c == 0:
        return (-b + math.sqrt(b**2-4*a*c)) / (2*a), #The extra comma turns output into tuple type, or else default will be a float
    
    else: #Otherwise return there are no solutions
        return None


#Prompt user inputs
x = float(input("Enter an a value:"))
y = float(input("Enter an b value:"))
z = float(input("Enter an c value:"))

#Print user inputs
print(f"Your a, b, c values are {x}, {y}, and {z}.")

#Run user inputs through the quad_roots function and define this as variable
solution = quad_roots(x,y,z)

#Check if there is no solution, if no solution, print no solution.
if solution == None:
    print("There are no real solutions to your quadratic equation.")

#If there is only 1 outputted number (length of tuple is 1), print the lone solution
elif len(solution) == 1: 
    print(f"The solution of your equation is {solution[0]}.") #solution[0] outputs the first value in the tuple. If entire solution is printed, it will include () and ,

else: #Otherwise output the 2 distinct solutions
    root_a, root_b = solution #Unpack the tuple output of the function into different variables root_a and root_b
    print(f"The solutions to your quadratic equation are {root_a} and {root_b}.")   
```
