## Problem 1 — Slicing practice
For `s = "calculus with python"`, print:
- First 8 characters
- Last 6 characters
- Every second character
- Reverse the string

```python

# TODO: Perform and print the described slices on s.
s = "calculus with python"
# Your code here...

#slices the text from the beginning until 8
s_first8 = s[:8]
print(s_first8)

#use negative slicing to print 6 characters starting from the end,
s_last6 = s[-6:]
print(s_last6)

#slicing with a step of 2 to print every other character
s_every2 = s[::2]
print(s_every2)

#use negative slicing to start from the end, until character #1
s_reverse = s[::-1]
print(s_reverse)
```

## Problem 3 — List comprehensions 

Construct  following, using list comprehension (without using loops).
- The list containing the first 100 odd numbers ($[1,3,5, ...]$).
- From `words = ["apple", "a", "banana", "pi", "grape"]`, create a new list of words, where all the words have length $\ge 3$

```python
# TODO: Implement the comprehensions.

#the formula for odd numbers is 2i + 1 for i starting at 0; this is included in the list and repeated 100 times
odds = [2 * i + 1 for i in range (100)]

words = ["apple", "a", "banana", "pi", "grape"]

#the word is included in the long_words list if:
#  the word is in the "words" list
#  the length of the word is ≥ 3
long_words = [word for word in words if len(word) >= 3]

#printing the results
print("The list has", len(odds), "numbers.")
print(odds)
print(long_words)
```
