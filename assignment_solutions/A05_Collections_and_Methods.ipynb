- Sort the list ascending
- Sort the list descending

```python
# TODO: Use list methods to transform nums as described.
nums = [3, 1, 2]
# Your code here...

#Append 5 to nums
nums.append(5)
print(nums)

#Extend 2,2,2 to nums
nums.extend([2,2,2])
print(nums)

#Remove 2 from nums
nums.remove(2)
print(nums)

#Count how many 2's are in nums
print(nums.count(2))
print(nums)

#Sorts nums from low to high
nums.sort()
print(nums)

#Reverse the numbers in num
nums.sort(reverse = True)
print(nums)

#Converts nums from a list to a tuple
nums = tuple(nums)
print(nums)
```
