```python
import numpy as np
#change the values below to test your code

#Input parameters. Adapt these to test your code.
max_iter = 1000
tolerance = 1e-10

#Prompting user to input which case we are testing
case_number = int(input("Enter the case number you want to test (1-4):"))

#Defining left and right boundaries a and b according to the case number
if case_number == 1:
    a = 1
    b = 10
elif case_number == 2:
    a = 0
    b = 2
elif case_number == 3:
    a = 3
    b = 4
elif case_number == 4:
    a = 5
    b = 7

#Defining f(x) according to the case number
def f(x):
    if case_number == 1:
        return np.exp(x/10) + np.log(x/10)
    elif case_number == 2:
        return np.arctan(x) - x**2
    elif case_number == 3:
        return np.sin(x)/np.log(x)
    elif case_number == 4:
        return np.log(np.cos(x))

#Whether an approximate root has been found
root_approximated = False

#Value of the approximate root (if found)
approximate_root = None

#Failure reason (if the algorithm fails)
bisection_failure = False
bisection_failure_reason = ""

#A variable to store the number of iterations
i = 0

# the left and right endpoint of the interval at each step
L = a
R = b

if f(L) * f(R) > 0:
    bisection_failure = True
    bisection_failure_reason = "bisection method does not apply: f(a) and f(b) have the same sign."

else:
    if f(L) == 0:
        approximate_root = L
        root_approximated = True

    elif f(R) == 0:
        approximate_root = R
        root_approximated = True

    while not root_approximated and i < max_iter:
        #implement the core logic

        # Computing the midpoint
        c = (L + R) / 2
        
        # Checking if f(midpoint) outputs 0; whether solution is found
        if f(c) == 0:
            approximate_root = c
            root_approximated = True

        # Checking if the interval size is smaller than the tolerance
        elif abs(R - L) < tolerance:
            approximate_root = c
            root_approximated = True

        # Shrinking the interval based on signs
        elif np.sign(f(c)) == np.sign(f(L)):
            L = c
        else:
            R = c

        i += 1  # increase the iteration tracker by 1 each time

    if not root_approximated:
        bisection_failure = True
        bisection_failure_reason = f"the algorithm reached the maximum iterations of {max_iter} without reaching the desired tolerance of {tolerance}."

#print whether or not the algorithm succeeded, and either the reason why it failed, or the approximate root
if bisection_failure:
    print(f"The algorithm was not able to find a root for Case {case_number} because the {bisection_failure_reason}")
else:
    print(f"Case {case_number}: The approximate root is {approximate_root}, and it took {i} iterations to find the root.")      
```
